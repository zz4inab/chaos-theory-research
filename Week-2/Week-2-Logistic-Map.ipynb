## **THE LOGISTIC MAP**
*Discovering fixed points and stability*

###**INTRODUCTION:**
The logistic map is a classic polynomial equation used in mathematics to model how populations change over generations. It is famous for demonstrating how complex, entirely unpredictable—or chaotic—behavior can arise from basic rules.
 It is therefore one of the most widely studied examples in dynamical systems and chaos theory.

The aim of this investigation is to determine the fixed points of the logistic map, analyse their stability using the derivative criterion, and derive analytically why the first bifurcation occurs at (r=3).






###**1. DEFINING THE LOGISTIC MAP:**
Defined by:

 $$x_{n+1}=rx_n(1-x_n),$$

 This equation describes how a sequence evolves through repeated iterations.
 Here, $x_n$ represents the current value of the sequence,
$x_{n+1}$ is the next value obtained after one iteration,
$r$ is a parameter that controls the behaviour of the system.

*   $x_n$ represents the current value of the system
*   $x_{n+1}$ is the next value obtained after an iteration
*   $r$ is a parameter controlling the systems behaviour (or growth rate in terms of populaions).

The equation is applied starting from $x_0$ to generate a sequence ie.
$$
x_0 \rightarrow x_1 \rightarrow x_2 \rightarrow x_3 \rightarrow \cdots
$$

As $r$ changes, the behavior of the system changes dramatically: For smaller valuess of $r$, such as 0.4, the sequences settles or converges to a single value, whereas larger values oscillate between 2, then 4, then 8 values before the behavior ultimately becomes chaotic and hard to describe.


### **2. FIXED POINTS:**
A fixed point is a value that remains unchanged when the logistic map is applied. If the system reaches a fixed point, every next iteration will produce the same value.

Mathematically, a fixed point satisfies

$$
f(x^*)=x^*.
$$
Which explains that when you put a fixed point value into the logistic map, it returns the same value.


Substituting the logistic map gives us

$$
x^*=rx^*(1-x^*).
$$

Rearranging and factorising eventually gives us two solutions:

The first solution is

$$
x^*=0,
$$

and the second is obtained by solving

$$
r(1-x^*)=1,
$$

which gives

$$
x^*=1-\frac1r.
$$

Therefore, the logistic map has two fixed points:

$$
x^*=0
$$

and

$$
x^*=1-\frac1r.
$$

### **3. STABILITY OF THE FIXED POINTS:**
Finding the fixed points alone is not sufficient. It is also necessary to determine whether nearby trajectories move towards or away from these points.

A fixed point is considered stable if trajectories that start very close to it gradually converge back to it. Conversely, it is unstable if nearby trajectories move further away after each iteration.

The stability of a fixed point is determined using the derivative of the map.

For the logistic map,

$$
f(x)=rx(1-x),
$$

the derivative is

$$
f'(x)=r(1-2x).
$$

A fixed point is stable if

$$
|f'(x^*)|<1.
$$

This criterion means that the map contracts nearby points rather than stretching them apart.

###**4. STABILITY CRITERION FOR EACH FIXED POINT:**
For the first fixed point, $x^*=0$, we have $f'(0)=r$. The stability condition $|r|<1$ gives $0<r<1$ for the usual logistic map range $0\le r\le 4$.

Thus, $x^*=0$ is stable only when $0<r<1$.

However, by substitutin the second fixed point, $(1-\frac1r)$, into the derivative, we obtain $f`(x^*)=2-r$.

According to the stability criterion, this becomes $|2-r|<1$, and solving this inequality ultimately gives us the final condition that is
**$$1<r<3$$**




### **5. THE FIRST BIFURCATION:**
for $1< r < 3$, the fixed point is stable as it fulfills the derivative condition for stability ie.    $$|f(`x)|<1$$
At $r=3$, the condition $|2-r|=1$ is still fulfilled, however $r$ is right at the boundary of stability, hence once $r$ becomes greater than 3, the **first bifurcation** occurs.

This means that for $r>3$, the values stop converging to a fixed point but instead osicillate between two values, this transition is called the first period doubling bifurcation. eventually this period-2 turns into period-4 where it oscillates between four values and so on.

**Numerical Verification:**


To verify the analytical results, the logistic map is simulated below for three values of the parameter $(r)$ using the same initial condition, ($x_0=0.2)$. For $(r=2.8)$, the sequence converged to a single fixed point, confirming that the fixed point is stable. At $(r=3.0)$, the convergence became much slower, indicating that the system is at the boundary of stability. For $(r=3.2)$, the sequence no longer converged to a single value but instead alternated between two values, forming a period-2 orbit. This agrees with the theoretical prediction that the first period-doubling bifurcation occurs at when $r>3$

*The code is inserted below*



```python
import matplotlib.pyplot as plt

def logistic_map(r, x0, n):
    x = x0
    orbit = [x]

    for _ in range(n):
        x = r * x * (1 - x)
        orbit.append(x)

    return orbit

# Parameters
x0 = 0.2
iterations = 30
r_values = [2.8, 3.0, 3.2]

plt.figure(figsize=(10, 6))

for r in r_values:
    orbit = logistic_map(r, x0, iterations)

    plt.plot(range(len(orbit)), orbit, marker='o', label=f"r = {r}")

    # Print the last 10 values
    print(f"\nLast 10 values for r = {r}:")
    print([round(x, 6) for x in orbit[-10:]])

plt.title("Logistic Map for Different Values of r")
plt.xlabel("Iteration")
plt.ylabel("x")
plt.legend()
plt.grid(True)

plt.show()
```

The logistic map demonstrates how complex behaviour can arise from a very simple nonlinear equation. By solving for the fixed points and analysing their stability using the derivative criterion, it is shown that the nonzero fixed point is stable only for $1<r<3$. At $r=3$, the fixed point loses stability and a stable period-2 orbit appears. This transition is the **first period-doubling bifurcation** and represents the beginning of the logistic map's transition from stability to chaos.

### **What is a Bifurcation Diagram?**

A bifurcation diagram is a graph that shows how the long-term behaviour of a dynamical system changes as a parameter is varied. For the logistic map,

$$
x_{n+1}=rx_n(1-x_n),
$$

the horizontal axis represents the parameter $r$, while the vertical axis shows the long-term values of $x$ after the initial transient has been discarded, which just show the sequence eventually settling to their fixed points. The diagram highlights the stable fixed points, periodic cycles, and chaotic patterns that the logistic map approaches for each value of $r$.

A single point indicates convergence to a stable fixed point, two points indicate a period-2 orbit, four points indicate a period-4 orbit, and further branching eventually leads to chaotic behaviour. The bifurcation diagram therefore provides a visual representation of the logistic map's route from stability to chaos through multiple period-doubling bifurcations.


```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MultipleLocator # Import MultipleLocator

# Initialize lists to store r and x values for plotting
r_plot_values = []
x_plot_values = []

# Range for parameter beta (equivalent to r in logistic map)
# Increased precision for r values for smoother diagrams
for beta in np.arange(0, 4.01, 0.001):
    xold = 0.5 # Initial condition

    # Flag to check if the system diverged for the current beta
    diverged = False

    # Discard first 2000 iterations (transient behavior)
    for _ in range(2000):
        # If xold leaves [0,1], the system diverged (e.g., due to r too high or initial x outside range)
        if not (0 <= xold <= 1):
            diverged = True
            break
        xnew = beta * xold * (1 - xold)
        xold = xnew

    if diverged:
        continue # Skip to the next beta if it diverged during transient

    # Collect the next 1000 iterations for the bifurcation diagram
    # These are the "long-term" values after the transient
    for _ in range(1000):
        # If xold leaves [0,1], stop collecting points for this beta
        if not (0 <= xold <= 1):
            break
        xnew = beta * xold * (1 - xold)
        xold = xnew
        r_plot_values.append(beta)
        x_plot_values.append(xnew)

# --- Plot 1: Full Range (0-4) ---
plt.figure(figsize=(12, 8))
plt.scatter(r_plot_values, x_plot_values, s=0.05, color='navy', alpha=0.8) # Changed to navy blue

ax1 = plt.gca()
ax1.set_facecolor('white') # Changed to white
ax1.tick_params(axis='x', colors='black') # Changed to black
ax1.tick_params(axis='y', colors='black') # Changed to black
ax1.set_xlabel("Parameter r", color='black') # Changed to black
ax1.set_ylabel("Long-term x values", color='black') # Changed to black
ax1.set_title("Bifurcation Diagram of the Logistic Map (Full Range)", color='black') # Changed to black
plt.xlim(0, 4) # Set x-axis limits for the full range
plt.ylim(0, 1) # Set y-axis limits to the valid range of x

fig1 = plt.gcf()
fig1.set_facecolor('white') # Changed to white
plt.grid(True, linestyle=':', alpha=0.6, color='darkgray') # Changed grid color
plt.show()

# --- Plot 2: Zoomed-in Range (3-4) ---
plt.figure(figsize=(12, 8))
plt.scatter(r_plot_values, x_plot_values, s=0.05, color='navy', alpha=0.8) # Changed to navy blue

ax2 = plt.gca()
ax2.set_facecolor('white') # Changed to white
ax2.tick_params(axis='x', colors='black') # Changed to black
ax2.tick_params(axis='y', colors='black') # Changed to black
ax2.set_xlabel("Parameter r", color='black') # Changed to black
ax2.set_ylabel("Long-term x values", color='black') # Changed to black
ax2.set_title("Bifurcation Diagram of the Logistic Map (Zoomed: r = 3 to 4)", color='black') # Changed to black
plt.xlim(3, 4) # Set x-axis limits to 3 to 4
plt.ylim(0, 1) # Set y-axis limits to the valid range of x
plt.xticks(np.arange(3, 4.1, 0.1)) # Set x-axis ticks to 0.1 intervals

# Add minor grid lines for 0.01 intervals
ax2.xaxis.set_minor_locator(MultipleLocator(0.01))
plt.grid(True, which='both', linestyle=':', alpha=0.6, color='darkgray') # Show both major and minor grids

fig2 = plt.gcf()
fig2.set_facecolor('white') # Changed to white
plt.show()
```

###**Estimating the Feigenbaum constant with these values**:
**What is the Feigenbaum Constant?**

The **Feigenbaum constant** $(\delta \approx 4.6692)$ is one of the most intriguing concepts in dynamical sytem. It describes the rate at which period-doubling bifurcations occur as the logistic map approaches chaos. A remarkable feature of the Feigenbaum constant is its universality: it appears in many seemingly unrelated nonlinear dynamical systems, demonstrating that the route to chaos follows a common mathematical pattern.

**Estimation and comparison:**
1. Calculating distances between the bifurcation points using this data:
$$d_0 = r_{n+1} - r_n = distance $$
$$d_1 = 3.449 - 3.000 = 0.449$$
$$d_2 = 3.544 - 3.449 = 0.095$$
$$d_3 = 3.565 - 3.544 = 0.021$$
$$d_4 = 3.570 - 3.565 = 0.005$$

2. Estimating the Feigenbaum constant with these values:


$$
\delta_1 = \frac{d_1}{d_2}
= \frac{0.449}{0.095}
\approx 4.726
$$


$$
\delta_2 = \frac{d_2}{d_3}
= \frac{0.095}{0.021}
\approx 4.524
$$


$$
\delta_2 = \frac{d_3}{d_4}
= \frac{0.021}{0.005}
\approx 4.2
$$


**Comparing the estimated and accepted values:**


|Estimate | Value|
|---|----------|
|$δ_1$ | 4.726 |
$δ_2$| 4.524|
$δ_3$| 4.2|
**Feigenbaum Constant**| 4.6692|


The estimated Feigenbaum constants are close to the accepted value of 4.6692. The differences are expected because the bifurcation points were estimated numerically from a simulation and are not the exact values of the points.











# **DAY 5: LYAPUNOV EXPONENTS:**

###**What is the Lyapunov Exponent?**
The **Lyapunov exponent** measures how quickly two trajectories that start very close together diverge over time in a dynamical system. It provides a means to determine whether a system is chaotic: a **positive** Lyapunov exponent indicates chaos, while a **negative** exponent indicates stable behaviour. Introduced by the Russian mathematician Aleksandr Lyapunov in the late 19th century, it is one of the most widely used tools for identifying chaotic behaviour in mathematics, physics, and engineering.

For the logistic map, the Lyapunov exponent is calculated using

$$
\lambda=\frac{1}{N}\sum_{n=1}^{N}\ln\left|r(1-2x_n)\right|,
$$

where (N) is the number of iterations. The value of λ provides a practical way to identify different types of behaviour, which is further explained below by overlying the logistic map and the lyapunov exponent for visualisation.

###**Computing the Lyapunov Exponent**

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------
# PARAMETERS
# ------------------------------------

r_values = np.linspace(2.5, 4.0, 1000)

iterations = 1000
transient = 800
x0 = 0.2

# ------------------------------------
# STORAGE
# ------------------------------------

bifurcation_r = []
bifurcation_x = []

lyapunov = []

# ------------------------------------
# MAIN LOOP
# ------------------------------------

for r in r_values:

    x = x0

    lyap_sum = 0

    for i in range(iterations):

        # Logistic map
        x = r * x * (1 - x)

        # Ignore transient behaviour
        if i >= transient:

            bifurcation_r.append(r)
            bifurcation_x.append(x)

            derivative = abs(r * (1 - 2*x))

            # Prevent log(0)
            derivative = max(derivative, 1e-12)

            lyap_sum += np.log(derivative)

    lyapunov.append(lyap_sum / (iterations - transient))

# ------------------------------------
# PLOT
# ------------------------------------

# Feigenbaum points for plotting
r_feigenbaum_points = [3.000, 3.449, 3.544, 3.565, 3.569]

fig, (ax1, ax2) = plt.subplots(
    2,
    1,
    figsize=(11.17,8.28),
    sharex=True,
    gridspec_kw={'height_ratios':[3,1]}
)

# Set a base font size for all text elements
plt.rcParams.update({'font.size': 14}) # Increased base font size

# Set figure and axes background colors to white
fig.set_facecolor('white')
ax1.set_facecolor('white')
ax2.set_facecolor('white')

# Bifurcation Diagram

ax1.plot(
    bifurcation_r,
    bifurcation_x,
    '.',
    color='#C39BD3', # Changed to a pinkish purple
    alpha=0.5,
    markersize=1 # Make points slightly larger for visibility with new color
)

ax1.set_ylabel("x", color='black', fontsize=16) # Set label color and increase font size
ax1.set_title("Logistic Map: Bifurcation Diagram with Lyapunov Exponent", color='black', fontsize=18) # Set title color and increase font size
ax1.tick_params(axis='x', colors='black', labelsize=14) # Set tick color and increase label size
ax1.tick_params(axis='y', colors='black', labelsize=14) # Set tick color and increase label size

# Add vertical lines for Feigenbaum constant calculation points
for r_point in r_feigenbaum_points:
    ax1.axvline(r_point, color='grey', linestyle='--', linewidth=0.8, alpha=0.7)

# Lyapunov Exponent

ax2.plot(
    r_values,
    lyapunov,
    color='maroon',
    linewidth=1
)

ax2.axhline(
    0,
    color='black',
    linestyle='--'
)

# Add vertical lines for Feigenbaum constant calculation points on ax2 as well
for r_point in r_feigenbaum_points:
    ax2.axvline(r_point, color='grey', linestyle='--', linewidth=0.8, alpha=0.7)

ax2.set_xlabel("Parameter r", color='black', fontsize=16) # Set label color and increase font size
ax2.set_ylabel("λ", color='black', fontsize=16) # Set label color and increase font size
ax2.tick_params(axis='x', colors='black', labelsize=14) # Set tick color and increase label size
ax2.tick_params(axis='y', colors='black', labelsize=14) # Set tick color and increase label size

plt.tight_layout()

# Save the figure to make it downloadable
plt.savefig('logistic_map_bifurcation_lyapunov.png', bbox_inches='tight', facecolor='white')

plt.show()
```

###**Analysis:**
By overlaying the two plots, we can infer what the values of λ indicate in terms of chaotic behavior

Corresponding to the graph; at a bifurcation point the value of λ reached zero, which shows how r is at the boundary of stability.

At stable regions we can infer that the value of λ is less than 1,and trajectories converge to a single fixed point, or oscillate between two values. This is the rea between bifurcations. At highly stable points such as 0 and 2, the value of λ approaches negative infinity (not visible in plotted graph).

Furhermore a value of λ which is positive indicates the divergence of trajectories and chaotic behavior which is seen after a value of approximately 3.57, as the value of λ crosses the x axis, and is seen to dip down to the negative region once in a while. This indicated stability windows before chaos is resumed again.

In conclusion, the Lyapunov Exponent is a way to redefine the mysterious chaotic regime in a mathematical manner and prove that it can be determined in hindsight whether or not a value of $r$ in a sequence will produce chaotic behavior.
